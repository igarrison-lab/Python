### Problem 1

#### Question Text

Write an Employee class that keeps data attributes for the following pieces of information:    

Employee name  
Employee number

Next, write a class named ProductionWorker that is a subclass of the Employee class. The ProductionWorker class should keep data attributes for the following information:    
Shift number (an integer, such as 1, 2, or 3)  
Hourly pay rate

The workday is divided into two shifts: day and night. The shift attribute will hold an integer value representing the shift that the employee works. The day shift is shift 1 and the night shift is shift 2. Write the appropriate accessor and mutator methods for each class.                            

Once you have written the classes, write a program that creates an object of the  class and prompts the user to enter data for each of the object’s data attributes. Store the data in the object, then use the object’s accessor methods to retrieve it and display it on the screen.

#### High-Level Overview

This one was actually not bad. The process of adding superclasses to subclasses is not such an abstract addition to what we had done previously. The logic made sense and is just creating another class and referencing that class. I learned the new syntax as it was also simple and made the two classes. From there, it was basically the same as before and I just used user input and print to display user attrivutes. Overall, I like.

![flowchart](zphotos/superclass.png)


```python
# add your code to this cell
# remember to include comments like this one for most of your code
# remember to "Run cell" once you're done, excecuting your code so the output is displayed below

# Create parent class
class Employee:
    # Initialize the attributes
    def __init__(self, name, employee_number):
        self.__name = name
        self.__employee_number = employee_number
    # Create the set methods
    def set_name(self, name):
        self.__name = name
    def set_employee_number(self, employee_number):
        self.__employee_number = employee_number
    # create the get methods
    def get_name(self):
        return self.__name
    def get_employee_number(self):
        return self.__employee_number

# create Procuction Worker subclass 
class ProductionWorker(Employee):
    # Initialize all attributes
    def __init__(self, name, employee_number, shift_number, pay_rate):
        # init parent attributes
        super().__init__(name, employee_number)
        # create new attributes for subclass
        self.__shift_number = shift_number
        self.__pay_rate = pay_rate
    # making methods for setting attribute
    def set_shift_number(self, shift_number):
        if shift_number == 1 or shift_number == 2:
            self.__shift_number = shift_number
        else:
            print("Please enter 1 or 2 as day or night shifts")
    def set_pay_rate(self, pay_rate):
        self.__pay_rate = pay_rate
    # returning the attributes safely
    def get_shift_number(self):
        return self.__shift_number
    def get_pay_rate(self):
        return self.__pay_rate

# Creating main function
def main():
    # Getting user input for their object
    user_name = input("Please enter the employee's name")
    user_employee_number = input("Please enter the employee's employee number")
    user_shift_number = int(input("Please enter the employee's shift number"))
    user_pay_rate = input("Please enter the employee's pay_rate")
    # creating user object with the inputs
    user_worker = ProductionWorker(user_name, user_employee_number, user_shift_number, user_pay_rate)
    # Displaying the attributes using the accessor methods
    print(f"Your name is {user_worker.get_name()}")
    print(f"Your employee number is {user_worker.get_employee_number()}")
    print(f"Your shift number is {user_worker.get_shift_number()}")
    print(f"Your pay rate is {user_worker.get_pay_rate()}")

# Calling the function
if __name__ == "__main__":
    main()
```

### Problem 2

#### Question Text

In a particular factory, a shift supervisor is a salaried employee who supervises a shift. In addition to a salary, the shift supervisor earns a yearly bonus when his or her shift meets production goals. Write a ShiftSupervisor class that is a subclass of the Employee class you created in Programming Exercise 1. The ShiftSupervisor class should keep a data attribute for the annual salary, and a data attribute for the annual production bonus that a shift supervisor has earned. Demonstrate the class by writing a program that uses a ShiftSupervisor object.

#### High-Level Overview

Going into this one, it wasn't looking hard at all. I simply used the previous answer format for the parent class and subclass making. For this one all I had to do was make the attributes which was much easier than the last so I thought to actually allow the worker to work and adding a bit extra. I just made the production attribute and added the work method to change the production attribute which was not hard just remembering the syntax and I got it all figured out.

```python
# add your code to this cell
# remember to include comments like this one for most of your code
# remember to "Run cell" once you're done, excecuting your code so the output is displayed below

# create parent employee class
class Employee:
    # Initialize the attributes
    def __init__(self, name, employee_number):
        self.__name = name
        self.__employee_number = employee_number
    # Create the set methods
    def set_name(self, name):
        self.__name = name
    def set_employee_number(self, employee_number):
        self.__employee_number = employee_number
    # create the get methods
    def get_name(self):
        return self.__name
    def get_employee_number(self):
        return self.__employee_number

# Creating the shift supervisor subclass
class ShiftSupervisor(Employee):
    # Initialize all attributes
    def __init__(self, name, employee_number, annual_salary, production_bonus, production):
        # init all parent attributes
        super().__init__(name, employee_number)
        # creating the subclass additional attributes
        self.__annual_salary = annual_salary
        self.__production_bonus = 0
        self.production = production
    # defining set methods for the attributes
    def set_annual_salary(self, annual_salary):
        self.__annual_salary = annual_salary
    def set_production_bonus(self, production_bonus):
        if self.production >= 100:
            self.__production_bonus = production_bonus
        else:
            print("Worker has not earned bonus yet")
    # defining get methods for the attributes
    def get_annual_salary(self):
        return self.__annual_salary
    def get_production_bonus(self):
        return self.__production_bonus
    # creating a work method for worker
    def work(self):
        self.production += 50
        return       
        
# Creating the supervisor object
my_shift_supervisor = ShiftSupervisor("Bill", 10, "100,000", 0, 0)
# supervisor working and earning bonus
for i in range(3):
    my_shift_supervisor.set_production_bonus(100000)
    my_shift_supervisor.work()
    print(f"suporvisor has earned {my_shift_supervisor.production} production")
    print(f"Supervisors current bonus amount is ${my_shift_supervisor.get_production_bonus()}")
    
```
