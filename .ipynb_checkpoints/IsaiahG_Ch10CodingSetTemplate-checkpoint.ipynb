### Problem 1

#### Question Text

Write a class named Pet, which should have the following data attributes:
__name(for the name of a pet)
__animal_type(for the type of animal that a pet is. Example values are ‘Dog’, ‘Cat’,and ‘Bird’)
__age(for the pet’s age)

The Pet class should have an __init__ method that creates these attributes. It should also have the following methods:
set_nameThis method assigns a value to the  field.
set_animal_typeThis method assigns a value to the  field.
set_ageThis method assigns a value to the  field.                                                                                                      get_nameThis method returns the value of the  field.
get_animal_typeThis method returns the value of the  field.
get_ageThis method returns the value of the  field.

Once you have written the class, write a program that creates an object of the class and prompts the user to enter the name, type, and age of his or her pet. This data should be stored as the object’s attributes. Use the object’s accessor methods to retrieve the pet’s name, type, and age and display this data on the screen. #### High-Level Overview

Starting off, the new terms and syntax was a bit tricky. I needed to constantly look back at the chapter for how to make the class.
```python
# add your code to this cell
# remember to include comments like this one for most of your code
# remember to "Run cell" once you're done, excecuting your code so the output is displayed below

# Making the class
class Pet:
    # Initializing the attributes
    def __init__(self, name, animal_type, age):
        self.__name = name
        self.__animal_type = animal_type
        self.__age = age
    # making the set methods
    def set_name(self, name):
        self.__name = name

    def set_animal_type(self, animal_type):
        self.__animal_type = animal_type
        
    def set_age(self, age):
        self.__age = age
# making the get accessor methods
    def get_name(self):
        print(f"Your pet's name is {self.__name}")

    def get_animal_type(self):
        print(f"Your pet's type is {self.__animal_type}")

    def get_age(self):
        print(f"Your pet's age is {self.__age}")

# Creating the Object instance
def user_pet():
    user_pet_name = input("What is your pet's name?:  ")
    user_pet_type = input("What is your pet's type?:  ")
    user_pet_age = float(input("What is your pet's age?:  "))
    my_pet = Pet(user_pet_name, user_pet_type, user_pet_age)
    # calling the info
    my_pet.get_name()
    my_pet.get_animal_type()
    my_pet.get_age()

# calling the function
if __name__ == "__main__":
    user_pet()
```

### Problem 2

#### Question Text

Write a class named Car that has the following data attributes:
__year_model(for the car’s year model)
__make(for the make of the car)
__speed(for the car’s current speed)

The Car class should have an __init__ method that accepts the car’s year model and make as arguments. These values should be assigned to the object’s __year_model and __make data attributes. It should also assign 0 to the __speed data attribute.

The class should also have the following methods:

- accelerate
The accelerate method should add 5 to the speed data attribute each time it is called.

- brake
The brake method should subtract 5 from the speed data attribute each time it is called.

- get_speed
The get_speed method should return the current speed.

Next, design a program that creates a Car object then calls the accelerate method five times. After each call to the accelerate method, get the current speed of the car and display it. Then call the brake method five times. After each call to the brake method, get the current speed of the car and display it.

#### High-Level Overview

*In two to three sentences at minimum, walk through your general approach to the problem and identify any particular issues you had.*

```python
# add your code to this cell
# remember to include comments like this one for most of your code
# remember to "Run cell" once you're done, excecuting your code so the output is displayed below

# Creating the car class
class Car:
    # initializing the attributes
    def __init__(self, year_model, make):
        self.__year_model = year_model
        self.__make = make
        self.__speed = 0
        
    # creating the methods
    def accelerate(self):
        self.__speed += 5
        return 

    def brake(self):
        self.__speed -= 5
        return 

    def get_speed(self):
        print(self.__speed)

# creating main function
def main():
    # creating the object
    my_car = Car("2014_Chevy", "Cruze")
    # making loops for the methods
    for i in range(5):
        my_car.accelerate()
        my_car.get_speed()
    for i in range(5):
        my_car.brake()
        my_car.get_speed()


# calling the main function
if __name__ == "__main__":
    main()
```
