#### Question Text

Write an Employee class that keeps data attributes for the following pieces of information:    

Employee name  
Employee number

Next, write a class named ProductionWorker that is a subclass of the Employee class. The ProductionWorker class should keep data attributes for the following information:    
Shift number (an integer, such as 1, 2, or 3)  
Hourly pay rate

The workday is divided into two shifts: day and night. The shift attribute will hold an integer value representing the shift that the employee works. The day shift is shift 1 and the night shift is shift 2. Write the appropriate accessor and mutator methods for each class.                            

Once you have written the classes, write a program that creates an object of the  class and prompts the user to enter data for each of the object’s data attributes. Store the data in the object, then use the object’s accessor methods to retrieve it and display it on the screen.

```python
# add your code to this cell
# remember to include comments like this one for most of your code
# remember to "Run cell" once you're done, excecuting your code so the output is displayed below

# Create parent class
class Employee:
    # Initialize the attributes
    def __init__(self, name, employee_number):
        self.__name = name
        self.__employee_number = employee_number
    # Create the set methods
    def set_name(self, name):
        self.__name = name
    def set_employee_number(self, employee_number):
        self.__employee_number = employee_number
    # create the get methods
    def get_name(self):
        return self.__name
    def get_employee_number(self):
        return self.__employee_number

# create Procuction Worker subclass 
class ProductionWorker(Employee):
    # Initialize all attributes
    def __init__(self, name, employee_number, shift_number, pay_rate):
        # init parent attributes
        super().__init__(name, employee_number)
        # create new attributes for subclass
        self.__shift_number = shift_number
        self.__pay_rate = pay_rate
    # making methods for setting attribute
    def set_shift_number(self, shift_number):
        if shift_number == 1 or shift_number == 2:
            self.__shift_number = shift_number
        else:
            print("Please enter 1 or 2 as day or night shifts")
    def set_pay_rate(self, pay_rate):
        self.__pay_rate = pay_rate
    # returning the attributes safely
    def get_shift_number(self):
        return self.__shift_number
    def get_pay_rate(self,):
        return self.__pay_rate

# Creating main function
def main():
    # Getting user input for their object
    user_name = input("Please enter the employee's name")
    user_employee_number = input("Please enter the employee's employee number")
    user_shift_number = int(input("Please enter the employee's shift number"))
    user_pay_rate = input("Please enter the employee's pay_rate")
    # creating user object with the inputs
    user_worker = ProductionWorker(user_name, user_employee_number, user_shift_number, user_pay_rate)
    # Displaying the attributes using the accessor methods
    print(user_worker.get_name())
    print(user_worker.get_employee_number())
    print(user_worker.get_shift_number())
    print(user_worker.get_pay_rate())

# Calling the function
if __name__ == "__main__":
    main()
```
